```python
import matplotlib.pyplot as plt
import numpy as np
import os

# Data
Voltage1 = np.array([0.431, 0.468, 0.511, 0.548, 0.573, 0.600, 0.654, 0.703, 0.753, 0.806, 
                     0.855, 0.900, 0.953, 1.002, 1.054, 1.103, 1.190, 1.252, 1.300, 1.343])
Current1 = np.array([0.02, 0.039, 0.071, 0.109, 0.135, 0.166, 0.235, 0.302, 0.375, 0.455, 
                     0.530, 0.600, 0.684, 0.764, 0.849, 0.929, 1.081, 1.180, 1.261, 1.340])

# Perform linear regression
m, c = np.polyfit(Voltage1, Current1, deg=1)
yfit = m * Voltage1 + c

# Create plot
plt.figure(figsize=(7, 5))
plt.scatter(Voltage1, Current1, color="firebrick", zorder=2, label="Data Points")
plt.plot(Voltage1, yfit, c="black", ls="--", label=f"Best Fit: I = {m:.4f}V + {c:.4f}")

# Labels & Formatting
plt.xlabel("Voltage (V)", fontsize=12)
plt.ylabel("Current (mA)", fontsize=12)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)

# Get current working directory and save the file there
current_folder = os.getcwd()  # Get current directory
file_path = os.path.join(current_folder, "box_BtoA.png")

plt.savefig(file_path, dpi=300, bbox_inches="tight")  # Save plot

# Show plot
plt.show()

# Print confirmation & best-fit equation
print(f"Plot saved as: {file_path}")
print(f"Best-fit equation: I = {m:.4f}V + {c:.4f}")

print(m)

```

```python
#Component BC current flowing from B ---> C
import numpy as np
import matplotlib.pyplot as plt

Voltage4 = np.array([0.409, 0.453, 0.507, 0.550, 0.601, 0.656, 0.701, 0.750, 0.804, 0.851, 0.901, 0.955, 1.012, 1.056, 1.108, 1.153, 1.205, 1.258, 1.306, 1.345])
Current4 = np.array([0.008, 0.023, 0.057, 0.096, 0.152, 0.221, 0.283, 0.353, 0.432, 0.501, 0.583, 0.668, 0.760, 0.832, 0.919, 0.992, 1.081, 1.169, 1.251, 1.317])  # mA

# Fit the data (linear regression)
m, c = np.polyfit(Voltage4, Current4, deg=1)

# Calculate the fitted values
yfit = m * Voltage4 + c

# Create best-fit equation string
equation_text = f"I = {m:.4f}V + {c:.4f}"

# Plot the fitted line and data points
plt.plot(Voltage4, yfit, c="black", linestyle="--", label=f"Best Fit: {equation_text}")
plt.scatter(Voltage4, Current4, color="firebrick", label="Data Points")

# Labels and grid
plt.xlabel("Voltage (V)")
plt.ylabel("Current (mA)")
plt.grid(True, linestyle="--", alpha=0.6)

# Display the legend with a slightly larger font
plt.legend(fontsize=10)

# Print the best-fit equation in the console
print(f"Best-fit equation: {equation_text}")

# Save the figure
plt.savefig("box_BtoC.png", dpi=300)

# Show the plot
plt.show()
```
