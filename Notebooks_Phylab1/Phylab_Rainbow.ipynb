```python
import numpy as np
import matplotlib.pyplot as plt
import math

def getcos(x):
    rad = math.radians(x)
    return math.cos(rad/6)

RIwater = 1.33
RIglyc = 1.47
```

# Water Orders

```python
water_readings = np.array([138.5083, 236.7083, 341.75, 479.0])
glycerine_readings = np.array([155.0, 261.5083, 111.525])
orders = [1,2,3,5]
#plt.plot(orders, water_readings)

plt.scatter(orders, water_readings, c="black", s=18)

coeffs = np.polyfit(orders, water_readings, deg=1)
slope, intercept = coeffs
linear_fit = np.poly1d(coeffs)

x_fit = np.linspace(min(orders), max(orders), 100)
y_fit = linear_fit(x_fit)

plt.scatter(orders, water_readings, c="black", s=18, label="Data")
plt.plot(x_fit, y_fit, c="red", label=f"Linear Fit\n$y = {slope:.2f}x + {intercept:.2f}$", ls="--")
plt.xlabel("Order")
plt.ylabel("Angle of Deviation Reading")
plt.legend()
plt.grid(True)
plt.show()
```

```python
phi1 = water_readings[1]
phi2 = glycerine_readings[1]
phis = [phi1, phi2]
cosdata = []
for i in phis:
    res = getcos(i)
    cosdata.append(res)

print(cosdata)

```

# Glycerine Orders

```python
glycerine_readings = np.array([155.0, 261.5083, 111.525+360])
ordersg = [1,2,3]
#plt.plot(orders, water_readings)

plt.scatter(ordersg, glycerine_readings, c="black", s=18)

coeffsg = np.polyfit(ordersg, glycerine_readings, deg=1)
m, c = coeffsg
linear_fitg = np.poly1d(coeffsg)

x_fitg = np.linspace(min(ordersg), max(ordersg), 100)
y_fitg = linear_fitg(x_fitg)

plt.scatter(ordersg, glycerine_readings, c="black", s=18, label="Data")
plt.plot(x_fitg, y_fitg, c="red", label=f"Linear Fit\n$y = {m:.2f}x + {c:.2f}$", ls="--")
plt.xlabel("Order")
plt.ylabel("Angle of Deviation Reading")
plt.legend()
plt.grid(True)
plt.show()
```

```python
multiplex = [0.7720, 0.6487, 1, 0.632]
multipley = [1/RIwater, 1/RIglyc, 1, 1/1.53]

coeffs = np.polyfit(multiplex, multipley, 2)
poly = np.poly1d(coeffs)

x_fit = np.linspace(min(multiplex), max(multiplex), 100)
y_fit = poly(x_fit)

eqn = f"$y = {coeffs[0]:.4f}x^2 + {coeffs[1]:.4f}x + {coeffs[2]:.4f}$"

plt.scatter(multiplex, multipley, color='black', label='Data Points')
plt.plot(x_fit, y_fit, color='red', label=eqn, ls="--")

plt.xlabel('cos(φ₂ / 6)')
plt.ylabel(r'1 / $\mu$')
plt.legend()
plt.grid(True)
plt.show()
```
