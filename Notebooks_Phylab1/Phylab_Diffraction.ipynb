# Part A - Wavelength Calculation

## Grating 1 - 100 lines per mm

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
x = np.array([1, 2, 3])

dist1 = 0.112
dist2 = 0.2265
dist3 = 0.341
avgd = np.array([dist1, dist2, dist3])
def get_sintheta(distances):
    thetas = []
    for i in distances:
        a = np.arctan(i / 2.042)
        b = math.sin(a)
        thetas.append(b)
    return np.array(thetas)

theta = get_sintheta(avgd)

d = 1 / 100000
dsintheta = np.array(d * theta)

plt.scatter(x, dsintheta, label="Data", c="darkblue")

m, b = np.polyfit(x, dsintheta, 1)
m_nm = m * 1e9

plt.plot(x, m*x + b, color='green', ls="--", label=f"Best Fit: y = ({m_nm:.3f} nm)x + {b:.5e} m")

plt.xlabel("Order of Diffraction (m)")
plt.ylabel("d * sin(θ) (m)")
plt.legend()
plt.grid()
plt.show()
```

## Grating 2 - 200 lines per mm

```python
x1 = np.array([1, 2, 3])

dist11 = 0.2275
dist21 = 0.4640
dist31 = 0.7185
avgd1 = np.array([dist11, dist21, dist31])

theta1 = get_sintheta(avgd1)

d1 = 1 / 200000
dsintheta1 = np.array(d1 * theta1)

plt.scatter(x1, dsintheta1, label="Data", c="darkblue")

m1, b1 = np.polyfit(x1, dsintheta1, 1)
m_nm1 = m1 * 1e9

plt.plot(x1, m1*x1 + b1, color='green', ls="--", label=f"Best Fit: y = ({m_nm1:.3f} nm)x1 + {b1:.5e} m1")

plt.xlabel("Order of Diffraction (n)")
plt.ylabel("d * sin(θ) (m)")
plt.legend()
plt.grid()
plt.show()
```

```python
print("The Wavelength obtained from grating of 100 lines per mm was found to be:  " + str(m) + " m")
```

## Grating 3 - 300 lines per mm

```python
x2 = np.array([1, 2, 3])

dist12 = 0.336
dist22 = 0.703
dist32 = 1.131
avgd2 = np.array([dist12, dist22, dist32])

theta2 = get_sintheta(avgd2)

d2 = 1 / 300000
dsintheta2 = np.array(d2 * theta2)

plt.scatter(x2, dsintheta2, label="Data", c="darkblue")

m2, b2 = np.polyfit(x2, dsintheta2, 1)
m_nm2 = m2 * 1e9

plt.plot(x2, m2*x2 + b2, color='green', ls="--", label=f"Best Fit: y = ({m_nm2:.3f} nm)x + {b2:.5e} m")

plt.xlabel("Order of Diffraction (n)")
plt.ylabel("d * sin(θ) (m)")
plt.legend()
plt.grid()
plt.show()
```

# Part B - Single Slit Width Calculation

```python
slitdist = np.array([0.0145, 0.0265, 0.03825])
mslit = np.array([1, 2, 3])
mslit1 = mslit * 546.542 / 1e9

sinarray = get_sintheta(slitdist)

plt.scatter(sinarray, mslit1, label="Data")

slope, intc = np.polyfit(sinarray, mslit1, deg=1)


plt.plot(sinarray, slope * sinarray + intc, color='black', ls="--", 
         label=f"Best Fit: y = ({slope:.6f} m) * sin(theta) + {intc:.2e}")

plt.xlabel("sin(theta)")
plt.ylabel("mλ (nm)")
plt.legend()
plt.grid()
plt.show()
print(slope)
```

# Part C - Diameter of a Circular Aperture

```python
circdists = np.array([0.005, 0.009, 0.015])
m_circ_first = np.array([1.22, 2.23, 3.23])
m_circ = m_circ_first*546.542 / 1e9

circsinarray = get_sintheta(circdists)

plt.scatter(circsinarray, m_circ, label="Data")

cslope, cintc = np.polyfit(circsinarray, m_circ, deg=1)


plt.plot(circsinarray, (cslope * circsinarray + cintc), color='black', ls="--", 
         label=f"Best Fit: y = ({cslope:.6f}m) * sin(theta) + {cintc:.2e}")

plt.xlabel(r"$sin(\theta)$")
plt.ylabel(r"$\overline{m}\ λ$ (nm)")
plt.legend()
plt.grid()
print(str(cslope) + " m")
```

# Part D - Single Helix Pitch and Thickness

```python
lambdaD = 2.042*548.8549 / 1e9
y_vals = np.array([lambdaD, 2*lambdaD, 3*lambdaD])
costheta = np.cos(0.436332)
delta2 = np.array([0.001*costheta, 0.00125*costheta, 0.00175*costheta])
plt.scatter(delta2, y_vals)

hslope, hintc = np.polyfit(delta2, y_vals, deg=1)


plt.plot(delta2, (hslope * delta2 + hintc), color='firebrick', ls="--", 
         label=f"Best Fit: y = ({hslope:.6f}m) * $\Delta_2$ + {hintc:.2e}")

plt.xlabel(r"$\Delta_2 cos(\theta)$")
plt.ylabel(r"$m\lambda D$")
plt.legend()
plt.grid()
print(hslope)
```
