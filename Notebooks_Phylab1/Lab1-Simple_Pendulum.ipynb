```python
import numpy as np
import matplotlib.pyplot as plt
```

**PART A**

```python
masses = [25.43, 30.73, 11.67, 8.9, 26.93, 35.9, 20.13] #masses M1 to M7
masses = np.array(masses)
print(masses)
```

```python
timesA = [1.08, 1.072, 1.068, 1.079, 1.102, 1.092, 1.101] #time readings for 10 oscillations of M1 to M7 at 10 degrees amplitude.
timesA = np.array(timesA)
print(timesA)
```

```python
plt.scatter(masses, timesA, c="darkblue", s=12)
plt.xlabel("Mass ($m$)")
plt.ylabel("Time period ($s$)")

# Further expanding the y-axis range
plt.ylim(0.5, 1.6)
plt.savefig("Massplot")
plt.show()
```

**PART B**

```python
lengths = [15, 20, 25, 30, 35, 40, 45, 50, 55, 60] #lengths L1 to L10
lengths = np.array(lengths)
print(lengths)
```

```python
timesB = [0.766, 0.885, 0.991, 1.101, 1.177, 1.262, 1.342, 1.421, 1.475, 1.547] #time readings for 10 lengths at 10 degree amplitude for 10 osc.
timesB = np.array(timesB)
print(timesB)
```

```python
plt.scatter(lengths, timesB, c="black", s=12, label="Data Points") 
plt.plot(lengths, timesB, c="firebrick", label="Trend Line")  


plt.xlabel('length(cm)')
plt.ylabel("Time period(s)")
plt.legend() 
plt.grid()

plt.savefig("Lengthplot")
```

**PART C**

```python
angles = [10, 20, 25, 30, 35, 40, 45, 50, 60, 70] #angles theta from 10 to 70
angles = np.array(angles)
print(angles)
```

```python
timesC = [1.068, 1.112, 1.114, 1.119, 1.122, 1.129, 1.133, 1.123, 1.176, 1.195] #time readings for thetas with constant l and 10 osc
timesC = np.array(timesC)
print(timesC)
```

```python
plt.scatter(angles, timesC, c="darkgreen", s=12)  # Scatter plot showing no relation
# plt.plot(angles, timesC, c="firebrick")  # Plotting for reference
plt.xlabel("Angle($^\circ$)")
plt.ylabel("Time period ($s$)")

# Expanding the y-axis range significantly to compress the points
plt.ylim(0.5, 1.7)

plt.savefig("Thetaplot")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Assuming timesB and lengths are already defined arrays
# Example:
# timesB = np.array([...])
# lengths = np.array([...])

# Step 1: Square the timesB array (T^2)
squared = timesB**2
length=lengths/100
# Step 2: Calculate the line of best fit
m, b = np.polyfit(squared, length, 1)  # Fit line to squared vs lengths

# Step 3: Scatter plot (T^2 vs l)
plt.scatter(squared, length, c="black", s=12, label="Data points")

# Step 4: Plot the best-fit line (using squared as x, lengths as y)
plt.plot(squared, m*squared + b, color='green', label=f"Best Fit: $l = {m:.2f}T^2 + {b:.2f}$")

# Step 5: Customize plot
plt.xlabel("$T^2$ (s$^2$)")  # X-axis: T^2 (squared time)
plt.ylabel("$l$ (m)")  # Y-axis: Length (l)
plt.legend()
plt.grid(True)

# Step 6: Show the plot
plt.savefig("gravitation_plot")
m
```
