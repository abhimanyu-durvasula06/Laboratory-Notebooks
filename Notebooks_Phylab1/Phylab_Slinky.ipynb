```python
import matplotlib.pyplot as plt
import numpy as np

initial_m = np.array([2.1, 4.4, 6.6, 10, 14.4, 16.6, 20.1, 24.4, 26.6, 30, 34.4, 40, 44.4, 49])
initial_l = np.array([95.7, 98.3, 100, 103, 106.4, 108.7 , 111.9, 115.5, 117.9, 121, 125, 129.3, 133.2, 138.5])
lengths = []
masses = []

for i in initial_m:
    masses.append(9.81*i / 1000)
    
for i in initial_l:
    lengths.append((i - 84.2) / 100)

lengths=np.array(lengths)
masses = np.array(masses)


#plt.plot(lengths, masses, c="black")
plt.scatter(lengths, masses, c="darkblue", label="Data Points")
plt.grid()

m,c = np.polyfit(lengths, masses,  deg=1)
ydata = m*lengths + c
plt.plot(lengths, ydata, ls="--", c='black', label = fr'm={m:.4f} Δx + {c:.4f}')
plt.xlabel("Extension $\Delta x$ (m)")
plt.ylabel("Force - mg $(N)$")
plt.legend()
#plt.xlim(2,70)

plt.savefig("Extforceplot")

```

```python
xhung = [0.046, 0.10, 0.152, 0.201, 0.251, 0.298, 0.343, 0.386, 0.430, 0.470, 0.509, 0.546, 0.582, 0.612, 0.645, 0.675, 0.704, 0.731, 0.757, 0.781, 0.800, 0.822, 0.841, 0.856, 0.873, 0.888, 0.9, 0.909, 0.918, 0.925, 0.931, 0.935, 0.939]
alts = []
for i in range(2,67,2):
    alts.append(i)
xhung = np.array(xhung)
alts = np.array(alts)
```

```python
xarray = []
for i in alts:
    xarray.append(i/66)

print(xarray)
```

```python
l = 0.052
x_vals = []
for i in range(2,67,2):
    varse = (i/66)*0.052
    x_vals.append(varse)
print(x_vals)
    
```

```python
y = [a - b for a, b in zip(xhung, x_vals)]
print(y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

coeffs = np.polyfit(xarray, y, 2)  
poly = np.poly1d(coeffs)

x_fit = np.linspace(min(xarray), max(xarray), 100)
y_fit = poly(x_fit)

plt.scatter(xarray, y, c="black", s=25, label="Data points")
plt.plot(x_fit, y_fit, color='royalblue', label=f"Best fit: {coeffs[0]:.3f}x² + {coeffs[1]:.3f}x + {coeffs[2]:.3f}")

plt.xlabel(r"Unstretched coordinate ($\frac{x}{l}$)")  
plt.ylabel(r"Extension $\xi$ (m)")
plt.xlim(-0.05,1.05)
plt.grid()
plt.legend()
plt.savefig("Stretchprofile")
```
