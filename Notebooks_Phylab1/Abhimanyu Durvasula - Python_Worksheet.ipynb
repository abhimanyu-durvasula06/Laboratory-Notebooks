### **Question 1:** ###

Load the provided data file and plot it.

```python
#Import the necessary libraries for loading the data and plotting
import numpy as np
import matplotlib.pyplot as plt #required libraries now imported
```

```python
#Load the given data, ensure that the data exists in the same directory from  where
# you are executing the code, otherwise you can specify the full-path in the filename.

myfile = "./Abhimanyu Durvasula - test_data.txt" #Accessing the file 
loaded_data = np.loadtxt(myfile , delimiter=',' , unpack=True) #handling the two comma separated values

#Note that this is a two dimensional data i.e. there are two columns. The first column is the 
#xdata and second is the ydata. If you just open the data file, you will notice that the columns are
#separated (called delimiter) by comma.


#Hint: To load a 2 dimensional data you can use the following command: 
# numpy.loadtxt(filename , delimiter=',' , unpack=True)
#The delimiter option is used to specify how the data columns are separated (comma in our case) 
#and unpack option unpacks the 2d data into two separate arrays.

xdata,ydata = loaded_data #Unpacking into two different arrays 
print(xdata)
print(ydata) #Displaying the two arrays
```

```python
#Plot the loaded data, decide whether a scatter plot or a line plot should be used.

plt.scatter(xdata, ydata, c="black", s=10) #choosing scatter since not very clear line can be fitted
plt.plot(xdata,ydata, c="firebrick") #line is still plotted to all data points for comparison
plt.title("Test Data Plot") #naming graph
plt.xlabel("x data (rad)")
plt.ylabel("y data") #labelling axes
plt.legend(["Test Data"] )  #setting up legend
plt.grid() #adding grid to decrease confusion

plt.show()


```

```python
#Make a guess as to what mathematical function best describes the the data and plot it
#over the raw data. Label legends appropriately.

#Function seems to be that of f(x) = cos(x)

angles = np.linspace(0, 4*np.pi, 100) #using numpy to generate angles within limits of 0 and 4pi
plt.plot(angles,np.cos(angles)) #plotting angles vs cos of the angles


plt.scatter(xdata, ydata, c="black", s=10)
plt.plot(xdata,ydata, c="firebrick")
plt.title("Test Data Plot")
plt.xlabel("x data / angle (rad)")
plt.ylabel("y data / cos(angle)") #reproducing out test data plot
plt.legend([r"$f(\theta) = \cos \theta$", "Test Data"] ) #adding new cos data to legend
plt.grid()


#Do not forget to label the axis, title the plot and label the legend


```

Define a function that takes in values in degrees and converts it to radians and returns the converted value.

This exercise is optional but worth attempting as it will be helpful for you to easily convert between degrees and radians which will come in very handy at times.

```python
def deg_to_rad(x): #name of function
    radians = (x*np.pi) / 180 #using degrees to rad formula
    return radians #returning new variable after processing

deg_to_rad(90)
```
