```python
rawtherms = np.array([3.87,3.2,2.7,2.34,2.1,2.21,2,1.81,1.66,1.57,1.44,1.32,1.23,1.08,0.96,0.87,0.81,0.75,0.67,0.65,0.55,0.49,0.45,0.42,0.4,0.38,0.35,0.32,0.3,0.28,0.26,0.25,0.23,0.22,0.21,0.2,0.18,0.17,0.16,0.15,0.14,0.14,0.13,0.12,0.11,0.1])
logtherms = np.log(rawtherms)
invTemps = 1/(Merctemps3 + 273.15)
plt.scatter(rawtherms,invTemps, label="Data", c="brown",s=18)
plt.grid()
plt.legend()
plt.xlabel(r"Resistance ($\mathrm{k}\Omega$)")
plt.ylabel(r"Inverse Temperature $(\mathrm{K}^{-1})$")

def model_func(x, a, b, c):
    return a + b*x + c*x**3

popt, pcov = cf(model_func, logtherms, invTemps)
print(popt, pcov)
a,b,c = popt

TS_fit = model_func(logtherms, a, b, c)
plt.plot(rawtherms,TS_fit, label= fr'Best Fit: y = {a:.2f}$x^3$ + {b:.2f}x - {abs(c):2f}', c="k", ls="--")
plt.legend()
plt.ticklabel_format(axis='y', style='sci', scilimits=(0,0))
plt.savefig("thermistortemps.jpg")
```

```python
residuals = invTemps - TS_fit
ss_res = np.sum(residuals**2)
ss_tot = np.sum((invTemps - np.mean(invTemps))**2)
r_squared = 1 - (ss_res / ss_tot)

print(f"R-squared: {r_squared:.5f}")
```

```python
#Three different containers of different area were used to plot cooling curves for water, of volume 150 mL.

def ThermConvert(arr):
    res = []
    a, b, c = popt
    for i in arr:
        x = np.log(i)   
        den = a + b*x + c*x**3
        t = 1/den
        res.append(t)
    return np.array(res)

def exp_func(x, a, b, c):
    return a * np.exp(-b * x) + c
```
