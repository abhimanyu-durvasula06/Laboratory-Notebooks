```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 120
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string 

def getC(n, consts):
    for i in range(n):
        if i >= 3:
            C_val = (consts[i-3])/(i*(i-1))
            consts.append(C_val)
    return consts
    

def plotairy(n, consts, xarr):
    ysum = 0
    ylist = []
    coeffs = getC(n, consts)
    for j in range(len(xarr)):
        val = xarr[j]
        for i in range(len(coeffs)):
            ysum += (val**i)*coeffs[i]
        ylist.append(ysum)
    plt.plot(xarr, ylist, label="Airy solution plot")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(alpha=0.3)
    plt.plot([], [], ' ', label=r"$C_0 = 1, C_1 = 1, C_2 = 0$")
    plt.legend()
```

```python
x = np.linspace(-15,10,400)
C = [1,1,0]
plotairy(200, consts=C, xarr = x)
plt.ylim(-20,30)
```
