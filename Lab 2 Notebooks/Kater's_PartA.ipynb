```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 200
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels

#Plot for time period vs length for a uniform rod. (Kater's Part A)
```

```python
lengths = np.array([2.5, 4.8, 7.5, 9.6, 14.7, 19.7, 24.6, 29.6, 34.7, 39.8, 44.8, 49.8])/100
times = np.array([42.733, 31.5233, 25.6133, 22.8433, 19.4199, 17.856, 17.0433, 16.720, 16.646, 16.7166, 16.98, 17.063])/10

lengths2 = -1*lengths
times2 = np.array([46.99, 33.243, 26.01, 23.056, 19.47, 17.77, 17.0633, 16.57, 16.636, 16.643, 16.93, 17.0933])/10

xlist = np.concatenate((lengths, lengths2))
ylist = np.concatenate((times, times2))
```

```python
plt.scatter(xlist, ylist, c="firebrick", s=20, label="Data")
plt.xlabel("Length (m)")
plt.ylabel("Time Period (s)")
plt.axvline(0, c='darkblue', ls='--', label="Centre of Mass")
plt.axhline(1.665, c='darkgreen', ls='--', label="Equal Time Period", lw=1)
plt.axvline(-39.8/100, c='orange', ls='--', lw=1, label="Corresponding Length")
plt.axvline(-29.6/100, c='orange', ls='--', lw=1)
plt.axvline(39.8/100, c='orange', ls='--', lw=1)
plt.axvline(29.6/100, c='orange', ls='--', lw=1)
plt.legend(prop={'size': 8}, loc='best', frameon=True, framealpha=0.8)
plt.grid()
plt.savefig("lengthtime_katers.jpg")
```
