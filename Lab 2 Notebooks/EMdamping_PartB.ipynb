```python
#Tracker Data Analysis
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 100
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels
```

## 200g weight

```python
file = np.loadtxt("200g_mag_PartB.txt", delimiter=',', skiprows=2)
print(file)
```

```python
t = []
y = []
for i in file:
    t.append(i[0])
    y.append(i[2])

yfull = y - y[0]
y = np.array(yfull[550:-100]/100)
t = np.array(t[550:-100])
```

```python
plt.scatter(t,y, c='firebrick', label='Data', s=2)

m,c = np.polyfit(t,y, deg=1)
y_fit = m*t + c
plt.plot(t, y_fit, c='darkblue', label='Best Fit')

plt.plot(t,y, c='k', lw=1)
plt.grid()
plt.legend()
plt.xlabel("time(s)")
plt.ylabel("y(cm)")
print(m)
```
