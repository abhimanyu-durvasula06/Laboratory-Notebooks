```python
#Tracker Data Analysis
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp

plt.rcParams['figure.dpi'] = 120
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels
```

## 200g weight

```python
file = np.loadtxt("200g_data_partA", delimiter=',', skiprows=2)
print(file)
```

```python
t = []
y = []
for i in file:
    t.append(i[0])
    y.append(i[1])

yfull = y - y[0]
y = np.array(yfull[100:400]/100)
t = np.array(t[100:400])
```

```python
plt.scatter(t,y, c='firebrick', label='Data', s=2)

a,b,c = np.polyfit(t,y, deg=2)
y_fit = a*t**2 + b*t + c
plt.plot(t, y_fit, c='darkblue', label='Best Fit')

# plt.plot(t,y, c='k', lw=1)
plt.grid()
plt.legend()
plt.xlabel("time(s)")
plt.ylabel("y(m)")
print(2*a)
```

## 250g weight

```python
file = np.loadtxt("250g_data_partA.txt", delimiter=',', skiprows=2)
print(file)
```

```python
t = []
y = []
for i in file:
    t.append(i[0])
    y.append(i[2])

yfull = y - y[0]
y = np.array(yfull[130:-25]/100)
t = np.array(t[130:-25])
```

```python
masses = np.array([50.7/1000, 101.1/1000, 151.6/1000, 201.9/1000, 252/1000])
accs = np.array([-0.004677, -0.01073, -0.01689, -0.02232, -0.02755])

m,c = np.polyfit(masses,accs, deg=1)
a_fit = m*masses + c
plt.plot(
    masses, a_fit, c="darkblue",
    label=fr"Best fit: $a = {m:.3f}\,m + {c:.7f}$"
)

plt.scatter(masses, accs, c='k', label= 'data')
plt.grid()
plt.xlabel("Mass (kg)")
plt.ylabel(r"Acceleration ($ms^{-1}$)")
plt.legend()
# plt.savefig("ma_plot.jpg")

sp.stats.linregress(masses, accs)
```
