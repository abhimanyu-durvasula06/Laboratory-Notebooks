```python
def findg(arr, dx, m, M):
    #Takes two time pairs in an array, has positional argument dx for length of object. Mass of rider is M, falling mass is m. Have to be given.
    a1, a2, b1, b2 = arr
    del1 = a2-a1
    del2 = b2-b1
    v1 = dx/del1
    v2 = dx/del2
    delV = v2-v1
    deltaT = b1-a2
    a = delV/deltaT
    g = a*( (m)/(m+M) )
    return g

def combmass(m_arr, M=0.2104):
    res_mass = []
    for i in m_arr:
        val = (i)/(i+M)
        res_mass.append(val)
    return np.array(res_mass)
        
     
```
