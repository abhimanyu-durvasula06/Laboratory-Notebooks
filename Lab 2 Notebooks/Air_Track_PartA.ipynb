```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 150
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels
```

```python
#Functions defined for analysis
def constv(arr, dx=1):
    #Takes two time pairs from photogate in array with positional argument dx which is length of the object. Returns ddlta v and percentage error in v.
    a1, a2, b1, b2 = arr
    del1 = a2-a1
    del2 = b2-b1
    v1 = dx/del1
    v2 = dx/del2
    delV = v2-v1
    print("Delta V :", delV)
    perc_err = delV/v1 * 100
    print("Percentage Error :", perc_err)

def allconstv(arr, dx):
    #Takes array of array of time pairs from photogate (multiple trials) with positional argument dx which is length of the object. Returns arrays of vel, delv, etc.
    V1s = []
    V2s = []
    Vdeltas = []
    percentages = []
    for i in arr:
        a1, a2, b1, b2 = i
        del1 = a2-a1
        del2 = b2-b1
        v1 = dx/del1
        v2 = dx/del2
        delV = v2-v1
        perc_err = delV/v1 * 100
        V1s.append(v1)
        V2s.append(v2)
        Vdeltas.append(delV)
        percentages.append(perc_err)
    return V1s, V2s, Vdeltas, percentages

def findg(arr, dx, m, M):
    #Takes two time pairs in an array, has positional argument dx for length of object. Mass of rider is M, falling mass is m. Have to be given.
    a1, a2, b1, b2 = arr
    del1 = a2-a1
    del2 = b2-b1
    v1 = dx/del1
    v2 = dx/del2
    delV = v2-v1
    deltaT = b2-a2
    a = delV/deltaT
    g = a*( (M+m)/m )
    return g
```
