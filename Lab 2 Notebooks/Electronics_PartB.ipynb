## RC-Integrator

```python
freq = np.array([100, 125, 150, 175, 200, 225, 250, 275, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750, 800, 850, 900, 950, 1000, 1050, 1100, 1150, 1200, 1250, 1300, 1350, 1400, 1450, 1500, 1550, 1600, 1800, 2000, 2200, 2400, 2600, 2800, 3000])
```

```python
v_in = np.concatenate((np.repeat(2.64, 35), np.repeat(2.56,7)))
```

```python
v_out = np.array([2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.64, 2.56, 2.48, 2.48, 2.48, 2.4, 2.32, 2.32, 2.32, 2.32, 2.24, 2.24, 2.24, 2.16, 2.16, 2.16, 2.08, 2, 2, 2, 2, 1.92, 1.92, 1.92, 1.76, 1.68, 1.6, 1.52, 1.44, 1.36, 1.28])
```

```python
gain = 20*np.log10(v_out/v_in)
logfreq = np.log10(freq)
```

```python
plt.title("Integrator Circuit")
plt.scatter(logfreq, gain, label="Data", c="firebrick", s=18)
plt.axvline(3.202, label = "Theoretical cutoff", c='darkgreen', ls='--')
plt.axvline(3.252, label = "Graph cutoff", c='k', ls='--')
# plt.plot(freqs, v_ratio, c='k')
plt.xlabel(r"$log_{10} \ Frequency \ (Hz)$")
plt.ylabel(r"Gain (dB)")
plt.legend()
plt.grid()
plt.savefig("bodeplotint.jpg")
```
