```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 120
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels
```

## LCR Series Circuit

```python
firstlist = np.arange (100, 1100, 100)
midlist = np.arange(1200, 1900, 200)
secondlist = np.arange (2000, 11000, 1000)
thirdlist = np.arange(20000, 60000, 10000)
fourthlist = np.arange(100000, 190000, 100000)
freqs = np.log10(np.concatenate((firstlist, midlist, secondlist, thirdlist, fourthlist)))
```

```python
v_in = np.concatenate((np.repeat(2.6, 3), np.repeat(2.56,7), np.repeat(2.48, 2), np.repeat(2.56, 7), np.repeat(2.56,1), np.repeat(2.52,1), np.repeat(2.48,2),np.repeat(2.64, 2), np.repeat(2.60, 3)))
```

```python
v_out = np.array([158e-3, 276e-3, 416e-3, 552e-3, 688e-3, 832e-3, 980e-3, 1.12, 1.26, 1.38, 1.56, 1.76, 1.88, 1.98, 2.12, 1.84, 1.48, 1.20, 960e-3, 840e-3, 720e-3, 640e-3, 560e-3, 352e-3, 240e-3, 184e-3, 144e-3, 80e-3])
```

```python
v_ratio = 20*np.log10(v_out/v_in)
```

```python
plt.title("LCR Series Circuit")
plt.scatter(freqs, v_ratio, label="Data", c="firebrick", s=20)
plt.axvline(3.3, ls='--', c='k', label='Graph peak')
plt.axvline(3.3117, ls='--', c='darkgreen', label='Theoretical peak')
plt.axhline(-3, ls='--', c="darkblue", label="Gain threshold")
plt.axvline(3.48, ls='--', c='orange', label=r'$f_2$')
plt.axvline(3.14, ls='--', c='pink', label=r'$f_1$')
# plt.plot(freqs, v_ratio, c='k')
plt.xlabel(r"$log_{10} \ Frequency \ (Hz)$")
plt.ylabel(r"Gain (dB)")
plt.legend(prop={'size': 8})
plt.grid()
plt.savefig("LCRseriesbodefinal.jpg")
```

## LCR Parallel Tank Circuit

```python
alist = np.array([1850, 1900, 1950, 2000])
firstlist = np.arange (100, 1100, 100)
midlist = np.arange(1200, 1900, 200)
seclist = np.arange(2100, 2600, 100)
secondlist = np.arange (3000, 11000, 1000)
thirdlist = np.arange(20000, 60000, 10000)

freq2 = np.log10(np.concatenate((firstlist, midlist, alist, seclist, secondlist, thirdlist, fourthlist)))
len(freq2)
```

```python
v_in = np.concatenate((np.repeat(2.48,12), np.repeat(2.56,1), np.repeat(2.64,10), np.repeat(2.48,2), np.repeat(2.32,1), np.repeat(2.48,10)))
len(v_in)
```
