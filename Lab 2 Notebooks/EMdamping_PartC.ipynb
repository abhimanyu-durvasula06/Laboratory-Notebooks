```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 120
plt.rcParams.update({'font.size':12, 'font.family':'serif'}) #Setting fonts
plt.rcParams['text.usetex'] = False

import string #For figure labels
```

```python
y = np.array([1393, 1305, 1238, 1169, 1106, 1054, 987, 934, 883, 837, 792, 751, 711, 675, 640])
x = np.array([9.52, 10.52, 11.52 ,12.52, 13.52, 14.52, 15.52, 16.52, 17.52, 18.52, 19.52, 20.52, 21.52, 22.52, 23.52])


plt.scatter(x,y, c='firebrick')
# plt.plot(x,y, c="k")


a, b, c = np.polyfit(x, y, deg=2)
y_fit = a*x**2 + b*x + c
equation = f"$y = {a:.2f}x^2 + {b:.2f}x + {c:.2f}$"
plt.plot(x, y_fit, c='darkblue', label=f'Best Fit: {equation}')

plt.xlabel("Magnet Separation (mm)")
plt.ylabel("Magnetic Field (Gauss)")
plt.grid()
plt.legend()
plt.savefig("sepvsgauss.jpg")
```
