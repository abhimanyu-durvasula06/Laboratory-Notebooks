```python
import astrolab
import numpy as np
import matplotlib.pyplot as plt
from astrolab import spectroscopy as spec
from astrolab import imaging as im
```

```python
image = im.load_image("./DSC_2534.JPG", print_log=True, stretch="linear")
```

```python
rot_img = im.rotate(image, angle=-5, print_log=True)
```

```python
croppedspec = im.crop(rot_img, left=1620, right=3040, top=1610, bottom=1230, print_log=True)
```

```python
plt.plot(croppedspec[200], color='firebrick', lw=0.75)
plt.plot(croppedspec[230], color='navy', lw=0.75)

plt.xlabel("Pixel")
plt.ylabel("Intensity")
```

```python
spectrum = spec.get_spectrum(croppedspec, print_log=True)
```

```python
wavelengths, AngstromsPerPix = spec.calibrate(spectrum, lineA=[535, 5890.0], lineB=[1033,4307.0], print_log=True)
```

```python
fig, ax = plt.subplots() # Create a blank canvas on which to plot both the spectrum and the reference lines

ax.plot(wavelengths, spectrum, color='k')  # Plot the spectrum
spec.plot_fraunhofer(fig=fig, ax=ax, lw=1) # Plot the Fraunhofer lines

plt.xlabel("Wavelength ($\AA$)")
plt.ylabel("Intensity")
plt.show()
```
