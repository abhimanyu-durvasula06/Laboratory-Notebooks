```python
#Code to get the Wavelength of spectral lines in Mercury Lamp Spectroscopy using angles in MSR and VSD from the spectroscope. 
import math
```

```python
def get_ang(angdict, m):
    anglelist = []
    zer = 81.16667
    for key, value in angdict.items():
        val = key + (20/3600)*value - zer
        sins = math.sin(math.radians(val))
        d = sins*(1/300)*10**(6)
        upd = d/m
        anglelist.append(round(upd, 5))
    return(anglelist)
    

order1ang = {88:35, 88.66:20, 89.66:0, 89.661:31, 90.66:0, 91.00:40, 91.001:50, 91.66:9}
order2ang = {95:50, 96.33:10, 98.33:11, 98.331:29, 100:50, 101.33:17, 101.33:17, 101.331:29}
order3ang = {104:50, 110.33:40, 112:58, 112.33:32}
angles1 = get_ang(order1ang, 1)
angles2 = get_ang(order2ang,2)
angles3 = get_ang(order3ang,3)

```

```python
print("Order 1:")
print(angles1)
print("Order 2:")
print(angles2)
print("Order 3:")
print(angles3)
```
